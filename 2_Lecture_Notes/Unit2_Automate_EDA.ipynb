## 🔹 2.1 Python Libraries to Automate Exploratory Data Analysis (EDA)

### **Introduction**

Exploratory Data Analysis (EDA) is the first and most important step in any data analysis or data science project.
It helps us understand the **structure, patterns, trends, and anomalies** present in a dataset before applying any machine learning or statistical model.

Python provides powerful libraries that **automate and simplify EDA tasks**, such as:

* Loading datasets
* Cleaning data
* Handling missing values
* Computing statistical summaries
* Understanding distributions and relationships

The two most fundamental libraries for automated EDA are **NumPy** and **Pandas**.

---

### **Why Python for EDA?**

Python is widely used for EDA because:

* It is easy to learn and readable
* It supports large datasets
* It has rich libraries for data handling and visualization
* It integrates well with machine learning frameworks

## 🔹 2.2 Regression

### **Meaning of Regression**

Regression is a statistical and analytical technique used to study the **relationship between variables** and to **predict the value of one variable based on another**.

It answers questions such as:

* How does one variable change when another changes?
* Can we predict future outcomes based on past data?

**Example:**
Predicting student marks based on hours of study.

---

### **2.2.1 Characteristics of Regression**

1. **Relationship Analysis**

   * Measures cause–effect relationship
   * One variable influences another

   📌 Example: Rainfall → Crop yield

2. **Direction of Relationship**

   * Positive: Both variables increase together
   * Negative: One increases, the other decreases

   📌 Example:

   * More study hours → Higher marks (positive)
   * Higher price → Lower demand (negative)

3. **Predictive Nature**

   * Used for prediction and forecasting
   * Helps in decision-making

   📌 Example: Predicting sales for the next month.

4. **Mathematical Model**

   * Regression creates a mathematical equation that represents the relationship.
   
Simple Linear Regression Equation:

   ```text
   y = a + bx
   ```

   where

   * `y` = dependent variable
   * `x` = independent variable
   * `a` = intercept
   * `b` = slope

5. **One-way Dependency**

   * Independent variable affects dependent variable
   * Unlike correlation, the dependency is not mutual.


---

### **2.2.2 Dependent and Independent Variables**

#### **Independent Variable (X)**

* Causes change
* Also called predictor or explanatory variable

Examples:

* Hours studied
* Temperature
* Advertising budget

#### **Dependent Variable (Y)**

* The variable that depends on X
* Also called response or target variable

Examples:

* Exam score
* Ice-cream sales
* Product demand

---

### **Simple Example Table**

```markdown
| Independent Variable (X) | Dependent Variable (Y) |
|--------------------------|------------------------|
| Hours of study           | Marks obtained         |
| Rainfall                 | Crop yield             |
| Experience (years)       | Salary                 |
```


📌 Key Point:
X influences Y, but Y does not influence X

---

## 🔹 2.2.3 Covariance and Correlation

### **A. Covariance**

Covariance measures how two variables change together.

**Interpretation**

* Positive → move in same direction
* Negative → move in opposite direction
* Zero → no linear relationship

📌 Example:
* Study hours & marks → Positive covariance
* Price & demand → Negative covariance

**Limitation**

* Covariance does not show strength of relationship
* Depends on the units of measurement

---

### **B. Correlation**

Correlation measures **direction and strength** of relationship.

📌 Correlation value ranges from –1 to +

| Value | Interpretation               |
| ----- | ---------------------------- |
| +1    | Perfect positive correlation |
| –1    | Perfect negative correlation |
| 0     | No correlation               |


📌 Example:
* Height & weight → Strong positive correlation
* Speed & travel time → Negative correlation
---

### **Difference Between Covariance and Correlation**

```markdown
| Feature            | Covariance | Correlation |
|--------------------|------------|-------------|
| Measures direction | ✔️         | ✔️          |
| Measures strength  | ❌         | ✔️          |
| Unit dependent     | ✔️         | ❌          |
| Range              | –∞ to +∞   | –1 to +1    |
```

#### Relation to Regression
* Correlation tells how strongly variables are related.
* Regression tells how much Y changes when X changes.

📌 Correlation explains association, regression explains prediction.

Summary:
* Regression predicts outcomes.
* Independent variable causes change.
* Dependent variable responds to change.
* Covariance shows direction.
* Correlation shows direction and strength.

---


### **Python Example: Regression**

Step 1: Import Libraries

```python
import numpy as np
import pandas as pd
```
Step 2: Create Dataset
```python
data = {
    'Hours_Studied': [1, 2, 3, 4, 5, 6],
    'Marks': [35, 40, 50, 55, 65, 70]
}

df = pd.DataFrame(data)
df
```
Out put of df
📌 Explanation:
• Hours_Studied → Independent Variable (X)
• Marks → Dependent Variable (Y)

Step 3: Separate Independent and Dependent
Variables
```python
X = df['Hours_Studied']
Y = df['Marks']
```
Step 4: Calculate Regression Line (Using NumPy)
```python
# Calculate slope (b) and intercept (a)
b, a = np.polyfit(X, Y, 1)
print("Slope (b):", b)
print("Intercept (a):", a)
```
output:


📌 Regression Equation:
Marks = a + b * Hours_Studied

Step 5: Predict Marks for New Data

```python
# Prediction
hours = 7
predicted_marks = a + b * hours
predicted_marks
```
Output:



📌 Interpretation:
“If a student studies for 7 hours, the model
predicts the marks.”

#### Python Examples using Pandas & NumPy for Covariance and Correlation

##### 🔹 Introduction

In data analytics, understanding the **relationship between variables** is essential.
Two important statistical measures used for this purpose are:

* **Covariance**
* **Correlation**

These measures help identify how variables move together and how strong their relationship is.

---

## 🔹 A. Covariance

### **Meaning of Covariance**

Covariance tells us **how two variables change together**.

It shows:

* Whether two variables **increase or decrease together**, or
* Whether **one increases while the other decreases**.

---

### **Interpretation of Covariance**

* **Positive Covariance** 👉 Both variables move in the same direction
  *(When X increases, Y also increases)*

* **Negative Covariance** 👉 Variables move in opposite directions
  *(When X increases, Y decreases)*

* **Zero Covariance** 👉 No clear linear relationship

---

### **Examples of Covariance**

* **Hours Studied and Marks Scored**

  * More study → More marks
  * Covariance is **positive**

* **Speed of a Vehicle and Time Taken**

  * More speed → Less time
  * Covariance is **negative**

---

### **Important Points about Covariance**

* Covariance shows **direction**, not strength
* Value depends on **units of measurement**
* Used in:

  * Statistics
  * Data analysis
  * Finance
  * Machine learning

---

### **Relation between Covariance and Correlation**

```markdown
| Covariance              | Correlation                     |
|-------------------------|---------------------------------|
| Shows direction         | Shows direction + strength      |
| Unbounded value         | Value between –1 and +1         |
| Unit dependent          | Unit free                       |
```

---

### **Definition**

> **Covariance is a statistical measure that indicates the direction of the linear relationship between two variables.**

---

### **Computing Covariance using Pandas**

```python
covariance = df['Hours_Studied'].cov(df['Marks'])
covariance
```


### **Computing Covariance using NumPy**

```python
import numpy as np

cov_matrix = np.cov(X, Y)
cov_matrix
```

---

### **Explanation of Covariance Matrix**

* **Diagonal values** → Variance
* **Off-diagonal values** → Covariance
* **Positive value** → Variables move in same direction

---

## 🔹 B. Correlation

### **Meaning of Correlation**

Correlation measures **both the direction and strength** of the relationship between two variables.

---

### **Computing Correlation using Pandas**

```python
correlation = df['Hours_Studied'].corr(df['Marks'])
correlation
```


---

### **Computing Correlation using NumPy**

```python
corr_matrix = np.corrcoef(X, Y)
corr_matrix
```

---

### **Interpretation of Correlation Values**

* Value close to **+1** → Strong positive correlation
* Value close to **–1** → Strong negative correlation
* Value close to **0** → Weak or no correlation

---

# 📘 **Unit 2.4 – Types of Machine Learning**

## 🔹 Introduction to Types of Machine Learning

Machine Learning (ML) is a subfield of **Artificial Intelligence (AI)** that focuses on building algorithms and models that allow computers to:

* Learn from data
* Improve with experience
* Make decisions without explicit programming for every task

### **In simple words**

> *Machine Learning teaches systems to learn patterns and make decisions like humans by analyzing data.*

There are **three main types of Machine Learning**:

1. Supervised Machine Learning
2. Unsupervised Machine Learning
3. Reinforcement Learning



---

## 🔹 1. Supervised Machine Learning

### **Definition**

Supervised Learning is defined as a learning method where a model is trained on a **labeled dataset**.

* Labeled data contains **input variables (features)** and **output variables (labels)**
* The algorithm learns a mapping between input and correct output
* Both training and validation datasets are labeled

---

### **Key Characteristics of Supervised Learning**

* Requires labeled data
* Learns from known outcomes
* Used for prediction tasks
* Suitable for both classification and regression problems



### **Two Main Categories of Supervised Learning**

---

## **1. Classification**

### **Meaning**

Classification predicts **categorical outputs**, meaning it assigns data points to predefined classes.

### **Examples**

* Spam vs Non-spam emails
* Disease present / not present
* Pass / Fail classification

---

### **Common Classification Algorithms**

* Logistic Regression
* Decision Tree
* Random Forest
* K-Nearest Neighbors (KNN)
* Naive Bayes
* Support Vector Machine (SVM)

---

## **2. Regression**

### **Meaning**

Regression predicts **continuous numerical values**.

### **Examples**

* House price prediction
* Sales forecasting
* Salary prediction

---

### **Common Regression Algorithms**

* Linear Regression
* Polynomial Regression
* Ridge Regression
* Lasso Regression
* Decision Tree
* Random Forest

---

### **Where to Use Supervised Learning**

* When labeled historical data is available
* When predicting outcomes is required
* Suitable for:

  * Classification problems
  * Regression problems

---

### **Applications of Supervised Learning**

* Image, speech, and text processing

For tasks like image classification, speech recognition and sentiment analysis.

* Predictive analytics (sales, churn, stock prices)

To forecast sales, customer churn, stock prices and weather conditions.

* Recommendation and personalization systems

Powering systems that suggest products, movies or content.

* Healthcare (diagnosis)
* Finance (fraud detection, credit scoring)

Used for medical diagnosis, fraud detection and credit scoring.

* Automation and control systems

In autonomous vehicles,manufacturing quality checks and gaming AI.

---

## 🔹 2. Unsupervised Machine Learning

### **Definition**

Unsupervised Learning works with **unlabeled data**, meaning there are no predefined outputs.

* The algorithm identifies **hidden patterns or structures**
* No training labels are provided
* Useful for exploratory data analysis

---

### **Key Characteristics**

* No labeled output
* Finds patterns automatically
* Used mainly for:

  * Clustering
  * Dimensionality reduction
  * Data visualization




### **Example**

In case of customer data without labels, the algorithm can group similar customers based on purchase behavior useful for segmentation and marketing

---

## 🔹 Categories of Unsupervised Learning

---

### **1. Clustering**

Clustering is the process of grouping data points into clusters based on their similarity. This technique is useful for identifying patterns and relationships in data without the need for labeled examples.

### **Common Clustering Algorithms**

* K-Means
* DBSCAN
* Mean-Shift

---

### **2. Dimensionality Reduction**

Reduces the number of features while preserving important information.

### **Common Techniques**

* Principal Component Analysis (PCA)
* Independent Component Analysis (ICA)

---

### **3. Association Rule Learning**

Finds relationships between items in large datasets. It identifies rules that indicate the presence of one item implies the presence of another item with a specific probability. 

### **Common Techniques**

* Apriori
* FP-Growth
* Eclat

---

### **Uses of Unsupervised Learning**

* When data is unlabeled or unstructured
* Useful for exploratory analysis, clustering or feature extraction.
* Common in marketing, recommendation systems and fraud detection where patterns matter more than labels.
* Feature extraction

---

### **Applications of Unsupervised Learning**

* Clustering and segmentation

Group similar data points, customers or images.

* Anomaly and fraud detection

Spot unusual patterns or outliers in data.

* Dimensionality reduction:

Simplify large datasets while retaining key information.

* Marketing analytics
* Customer behavior analysis

Identify user preferences and improve product suggestions.

* Data preprocessing and analysis

Clean data, detect patterns and support exploratory data analysis (EDA).

* Goal is to maximize total reward

### **Examples**
