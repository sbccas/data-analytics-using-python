## 1. Nikhilam Navatashcaramam Dashatah

```
def nikhilam_subtract(base, number):
    return base - number

nikhilam_subtract(1000, 357)
```

## 2. Ekadhikena Purvena

```
def square_ending_5(n):
    x = n // 10
    return x * (x + 1) * 100 + 25

square_ending_5(35)
```

## 10. Yavadunam

```
def yavadunam(n, base=100):
    d = base - n
    return (n - d) * base + d**2

yavadunam(97)
```
