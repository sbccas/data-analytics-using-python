## 1. Nikhilam Navatashcaramam Dashatah
**Meaning:** All from 9 and last from 10

This sutra is used for subtraction from powers of 10 such as 10, 100, 1000.


```python
def nikhilam_subtract(base, number):
    return base - number

nikhilam_subtract(1000, 357)
```

## 2. Ekadhikena Purvena
Used to square numbers ending with 5.
Formula:
If number = x5 → Result = (x × (x+1)) followed by 25


```python
def square_ending_5(n):
    x = n // 10
    return x * (x + 1) * 100 + 25

square_ending_5(35)
```

## 4. Paraavartya
Used for division using remainder.

```python
def paraavartya_divide(dividend, divisor):
    return dividend // divisor, dividend % divisor

q, r = paraavartya_divide(458, 9)
print("Quotient:", q, "Remainder:", r)
```

## 5. Shunyam Saamyasa-muccaye
If A + B = C + D, then A − C = D − B


```python
def shunyam_sutra(a, b, c, d):
    return (a + b) == (c + d)

print("Check:", shunyam_sutra(2, 5, 1, 6))
```

```python
(2 + 5) == (1 + 6)
```

## 6. Anurupyena
Used in ratio and proportion problems.


```python
def anurupyena(cost, items, new_items):
    return (cost / items) * new_items

print("Cost for 6 items =", anurupyena(40, 4, 6))
```

```python
(40 / 4) * 6
```

## 7. Sankalana–Vyavakalan-abhyam
Problems solved using addition and subtraction.


```python
def sankalana(a, b):
    return a + b, a - b

print("Addition and Subtraction:", sankalana(17, 9))
```

```python
17 + 9, 17 - 9
```

## 8. Purana–Apurana–Abhyam
Used for completing the square.


```python
def complete_square(a, b):
    return f"(x + {b//2})² - {(b//2)**2}"

print(complete_square(1, 4))
```

```python
'(x + 2)^2 - 4'
```

## 9. Chalana–Kalana
(a − b) × c = (a × c) − (b × c)


```python
def chalana(a, b, c):
    return (a - b) * c

print("(100 − 1) × 6 =", chalana(100, 1, 6))
```

```python
(100 - 1) * 6
```

## 10. Yavadunam
Used to square numbers close to base (100).


```python
def yavadunam_square(n, base=100):
    d = base - n
    return (n - d) * base + d**2

print("97² =", yavadunam_square(97))
```

## 11. Vyastisamanstih
Used for fraction calculations.


```python
def part_whole(part, whole):
    return part / whole

print("80/200 =", part_whole(80, 200))
```

```python
80 / 200
```

## 12. Sesanyan
Used to find remainder.


```python
def sesanyan(dividend, divisor):
    return dividend % divisor

print("29 mod 6 =", sesanyan(29, 6))
```

```python
29 % 6
```

## 13. Gunitasamuchyah
Used to verify multiplication using digit sums.


```python
def digit_sum(n):
    return sum(map(int, str(n)))

print("Digit Sum Check:", digit_sum(32) * digit_sum(14), digit_sum(448))
```

```python
sum(map(int,str(32))) * sum(map(int,str(14))), sum(map(int,str(448)))
```

## 14. Vistaran
Used for algebraic expansion.


```python
def expand(a, b):
    return f"x² + {a+b}x + {a*b}"

print(expand(4, 6))
```

```python
'x^2 + 10x + 24'
```

## 15. Rupan
Used to convert fractions.


```python
def rupan(n, d):
    return n / d

print("3/4 =", rupan(3, 4))
```

```python
3/4
```

## 16. Chidana
Used to split numbers for multiplication.


```python
def chidana(a, b):
    return (a // 10 * 10 + a % 10) * b

print("18 × 7 =", chidana(18, 7))
```

```python
18 * 7
```
