```python
plt.hist(stud['reading score'],bins=10,edgecolor='black')
```

```python
# HISTOGRAM ON STUDENT SCORE MATHS, READING, WRITING
# plt.hist(df['reading score'],bins=20,edgecolor='red')
# plt.show
plt.hist(stud['math score'],bins=10,color='skyblue',edgecolor='black')
plt.title('Univariate Analysis of Math Score')
plt.xlabel('math score')
plt.ylabel('Frequency (No of Students)')
plt.show
# plt.hist(df['math score'],bins=20,edgecolor='red')
# plt.show
```

```python
plt.hist(stud['reading score'],bins=10,color='skyblue',edgecolor='black')
plt.title('Univariate Analysis of Reading Score')
plt.xlabel('Reading score')
plt.ylabel('Frequency (No of Students)')
plt.show
```
